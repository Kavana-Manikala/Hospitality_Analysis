## Trend Analysis

### Booking By Month 

```python
Monthly_Booking = Hospital_data['booking_date'].dt.strftime('%B').value_counts().reset_index()
Monthly_Booking.columns = ['Month', 'Total_Booking'] 
Monthly_Booking = Monthly_Booking.sort_values('Total_Booking',ascending=False)
print(Monthly_Booking)

'''
brief about    dt.strftime(%B)

dt=datetime
strftime=string format time
%B = gives full month name 

%b= gives abbrivated month name 

%m = gives month number

eg:
dt.strftime('%B') January
dt.strftime('%b'): Jan
dt.strftime('%m'): 01

'''
```

## Bar chart representation of Booking By Month

```python
plt.figure(figsize=(12,8))
plt.bar(Monthly_Booking['Month'],Monthly_Booking['Total_Booking'],color=plt.cm.tab20(np.arange(len(Monthly_Booking))))
for i, value in enumerate(Monthly_Booking['Total_Booking']):
    plt.text(i,value,str(value),ha='center',va='bottom')
plt.title('Total Booking by Month', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Revenue By Month

```python
Monthly_Revenue = Hospital_data.groupby(Hospital_data['check_in_date'].dt.strftime('%B'))['revenue_realized'].sum().reset_index()
Monthly_Revenue.columns = ['Month', 'Total_revenue'] 
Monthly_Revenue = Monthly_Revenue.sort_values('Total_revenue',ascending=True)
print(Monthly_Revenue)
```

## Bar chart representation of Revenue By Month

```python
plt.figure(figsize=(12,8))
plt.bar(Monthly_Revenue['Month'],Monthly_Revenue['Total_revenue'],color=plt.cm.tab20c(np.arange(len(Monthly_Revenue))))
for i, value in enumerate(Monthly_Revenue['Total_revenue']):
    plt.text(i,value,str(value),ha='center',va='bottom')
plt.title('Total Revenue by Month', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Weekday Weekend Analysis

### Weekday vs Weekend Revenue
